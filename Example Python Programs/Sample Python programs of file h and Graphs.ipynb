### Write a python program to perform the following operation for addition, subtraction, multiplication, division and modulo using function. (17)

```python
def addition(a, b):
    return a + b

def subtraction(a, b):
    return a - b

def multiplication(a, b):
    return a * b

def division(a, b):
    return a / b

def modulo(a, b):
    return a % b

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print("Addition =", addition(num1, num2))
print("Subtraction =", subtraction(num1, num2))
print("Multiplication =", multiplication(num1, num2))

if num2 != 0:
    print("Division =", division(num1, num2))
    print("Modulo =", modulo(num1, num2))
else:
    print("Division and modulo are not possible because the second number is 0.")
```

### Write a python program to print the length of each line in the file. (18)

```python
with open("te_file.txt", "r") as file:

    line_number = 1

    for line in file:

        length = len(line.rstrip("\n"))

        print("Length of line", line_number, "=", length)

        line_number += 1
```

### Draw a graph which satisfied following condition. (20)
- x = range (10)
- y = cos(x)
- graph should be “dot-dashed”
- Give the label to x-axis “this is X dimension”
- Give the label to y-axis “this is Y dimension”
- Give the Title to graph “This is My First Graph“

```python
import numpy as np
import matplotlib.pyplot as plt

x = range(10)

y = np.cos(x)

plt.plot(
    x, y,
    linestyle="-."
)

plt.xlabel("this is X dimension")
plt.ylabel("this is Y dimension")
plt.title("This is My First Graph")

plt.grid()
plt.show()
```

### Draw a graph which satisfies the following condition given practicality. (21)
- Give the limit to x-axis (-2,12)
- Give the limit to y-axis (-2,2)
- c)	        Display Annotation


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 11, 1)

y = np.sin(x)

plt.plot(
    x, y,
    marker="o"
)

plt.xlim(-2, 12)

plt.ylim(-2, 2)

plt.annotate(
    "Maximum Point",
    xy=(8, np.sin(8)),
    xytext=(5, 1.5),
    arrowprops=dict(
        arrowstyle="->"
    )
)

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Graph with Annotation")

plt.grid()
plt.show()
```

### Plot Line plot with formatting to display two student’s data with last 4 semester SPI. also use legend in plot , different line styles and marker characters. (22)

```python
import matplotlib.pyplot as plt

semester = [1, 2, 3, 4]
student_1_spi = [7.2, 7.8, 8.1, 8.5]
student_2_spi = [6.8, 7.4, 7.9, 8.2]

plt.plot(
    semester, student_1_spi,
    linestyle="-",
    marker="o",
    label="Student 1"
)

plt.plot(
    semester, student_2_spi,
    linestyle="--",
    marker="s",
    label="Student 2"
)

plt.xlabel("Semester")
plt.ylabel("SPI")
plt.title("SPI Performance of Two Students")
plt.xticks(semester)

plt.legend()
plt.grid()
plt.show()
```

### Write a python program to do line plotting, scatter plotting, Bar chart, box plot , histogram and pie chart. (23)

```python
import matplotlib.pyplot as plt

# Common data
x = [1, 2, 3, 4, 5]

y = [10, 25, 15, 30, 20]

# 1. Line Plot

plt.figure()

plt.plot(x,y,marker="o")

plt.title("Line Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.grid()
plt.show()
```

```python
# 2. Scatter Plot

plt.figure()

plt.scatter(x,y)

plt.title("Scatter Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.grid()
plt.show()
```

```python
# 3. Bar Chart

students = ["A","B","C","D"]

marks = [75, 82, 68, 90]

plt.figure()

plt.bar(
    students,
    marks
)

plt.title("Bar Chart")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
# 4. Box Plot

data = [10,15,18,20,22,25,30,35,40]

plt.figure()
plt.boxplot(data)
plt.title("Box Plot")
plt.ylabel("Values")
plt.show()
```

```python
# 5. Histogram

marks = [45, 50, 55, 60, 62, 65, 70, 72, 75, 80, 85, 90]

plt.figure()

plt.hist(marks,bins=5)

plt.title("Histogram")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

```python
# 6. Pie Chart

subjects = ["Python", "C", "Java", "DBMS"]

students_count = [30, 25, 20, 25]

plt.figure()

plt.pie(
    students_count,
    labels=subjects,
    autopct="%1.1f%%"
)

plt.title("Pie Chart")
plt.show()
```

### Write Python program to plot time series with trends of sales. (24)


```python
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]

sales = [120,135,150,145,170,185,200,210,205,225,240,260]

plt.plot(
    months, sales,
    marker="o",
    linestyle="-",
    label="Monthly Sales"
)

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales Trend")

plt.legend()
plt.grid()
plt.show()
```
