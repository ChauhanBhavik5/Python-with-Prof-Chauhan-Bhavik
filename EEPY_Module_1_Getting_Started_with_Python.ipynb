## Variables
### What is a Variable?

A variable is a named memory location used to store data.

##### Rules for Naming Variables

✔ Must begin with a letter or underscore

✔ Cannot start with a number

✔ Cannot contain spaces

✔ Cannot use keywords

#### Example:

```python
name = "Prof Chauhan Bhavik"
age = 30.5
salary = 88700

print(name)
print(age)
print(salary)
```

## type() Function

### Let's find area of ractangle using python

### Demonstrate use of type() and id() function with different data types of Python.

```python
a = 10
b = 10.5
c = "Python"
d = True

print("Value:", a, "Type:", type(a), "ID:", id(a))
print("Value:", b, "Type:", type(b), "ID:", id(b))
print("Value:", c, "Type:", type(c), "ID:", id(c))
print("Value:", d, "Type:", type(d), "ID:", id(d))
```
