A calculator is one of the simplest applications used to understand arithmetic operations in Python. Python provides built-in arithmetic operators that can perform various mathematical calculations on numeric data.

The input() function is used to accept values from the user. Since the input is received as a string, it must be converted into an integer or floating-point number using int() or float() before performing arithmetic operations.

## Arithmetic Operators in Python
Operator	     Description	       Example
-  "+"	         Addition	           10 + 5 = 15
-  "-"	         Subtraction	       10 - 5 = 5
-  "*"	         Multiplication	       10 × 5 = 50
-  "/"	         Division	           10 / 5 = 2.0
-  "//"	         Floor Division	       10 // 3 = 3
-  "%"	         Modulus (Remainder)   10 % 3 = 1
-  "**"	         Exponentiation	       2 ** 3 = 8

### Algorithm
1) Start the program.
2) Accept the first number from the user.
3) Accept the second number from the user.
4) Perform addition.
5) Perform subtraction.
6) Perform multiplication.
7) Perform division.
8) Perform floor division.
9) Perform modulus operation.
10) Perform exponentiation.
11) Display all the results.
12) Stop the program.


```python
# Simple Calculator using Arithmetic Operators

# Taking input from the user
num1 = float(input("Enter First Number: "))
num2 = float(input("Enter Second Number: "))

print("\n========== CALCULATOR ==========")

print("Addition        :", num1 + num2)
print("Subtraction     :", num1 - num2)
print("Multiplication  :", num1 * num2)

if num2 != 0:
    print("Division        :", num1 / num2)
    print("Floor Division  :", num1 // num2)
    print("Modulus         :", num1 % num2)
else:
    print("Division        : Not Possible (Division by Zero)")
    print("Floor Division  : Not Possible")
    print("Modulus         : Not Possible")

print("Exponentiation  :", num1 ** num2)
```

## Program Explanation
- The input() function accepts two numbers from the user.
- The float() function converts the input values into floating-point numbers.
- Arithmetic operators perform the required mathematical operations.
- The if statement checks whether the second number is zero before performing division, floor division, and modulus operations to avoid runtime errors.
- The calculated results are displayed using the print() function.
