# Class & Object Concepts
## Concept

### Class:
A class is a blueprint or template used to create objects. It defines the properties (data) and methods (functions) that an object can have.

### Object:
An object is an instance of a class. It can access the properties and methods defined inside the class.

In Python, a class is created using the class keyword. The __init__() method is commonly used to initialize object data.

#### Example:
If Student is a class, then student1 and student2 can be objects of the Student class.

#### Key Points:

- Class → Blueprint
- Object → Instance of a class
- Attribute → Data stored in an object
- Method → Function defined inside a class
- self → Refers to the current object

# 32: Demonstrate Class and Object Concepts

### AIM: Write a Python program to demonstrate class and object concepts.

```python
# Creating a class named Student
class Student:

    # This is a method used to initialize the object
    def __init__(self, name, age):

        # Store the name in the object
        self.name = name

        # Store the age in the object
        self.age = age

    # Creating a method to display student details
    def display(self):

        # Print the student's name
        print("Name:", self.name)

        # Print the student's age
        print("Age:", self.age)


# Creating the first object of the Student class
student1 = Student("Rahul", 20)

# Creating the second object of the Student class
student2 = Student("Priya", 21)


# Display details of the first student
student1.display()

# Display details of the second student
student2.display()
```

# Inheritance
## Concept

Inheritance is an Object-Oriented Programming concept in which one class can acquire the properties and methods of another class.

The existing class is called the parent class or base class, while the new class is called the child class or derived class.

Inheritance helps us reuse existing code and avoid writing the same code again.

#### Example:
If Animal is a parent class and Dog is a child class, the Dog class can use methods defined in the Animal class.

#### Key Points:

- Parent class → Base class
- Child class → Derived class
- Child class can use parent class methods
- Inheritance promotes code reusability

# 33 Inheritance

### Aim: Write a Python program that illustrates inheritance concepts.

```python
# Creating a parent class
class Animal:

    # Creating a method in the parent class
    def eat(self):

        # Print a message
        print("Animal is eating")


# Creating a child class
# Dog inherits the properties and methods of Animal
class Dog(Animal):

    # Creating a method for the Dog class
    def bark(self):

        # Print a message
        print("Dog is barking")


# Creating an object of the Dog class
dog1 = Dog()


# Calling the method inherited from the Animal class
dog1.eat()

# Calling the method of the Dog class
dog1.bark()
```

- Animal is the parent class.
- Dog is the child class.
- Dog inherits the eat() method from Animal.
- Therefore, the Dog object can use both eat() and bark().

# Polymorphism

## Concept

Polymorphism means "many forms." It allows the same method or operation to behave differently depending on the object or situation.

For example, different classes can have a method with the same name, but each class can provide its own implementation of that method.

In our example, both Dog and Cat have a sound() method, but they produce different sounds.

#### Key Points:

- Polymorphism means many forms.
- The same method name can have different behaviour.
- It makes programs more flexible and reusable.
- Different objects can respond differently to the same method call.

### 34 Aim: Write a Python program that illustrates the polymorphism concept.

```python
# Creating the first class
class Dog:

    # Creating a method with the same name
    def sound(self):

        # Print the sound of a dog
        print("Dog says: Woof")


# Creating the second class
class Cat:

    # Creating a method with the same name
    def sound(self):

        # Print the sound of a cat
        print("Cat says: Meow")


# Creating an object of Dog class
dog1 = Dog()

# Creating an object of Cat class
cat1 = Cat()


# Calling sound() method using Dog object
dog1.sound()

# Calling sound() method using Cat object
cat1.sound()
```

# Searching Algorithms

## Concept

Searching is the process of finding a particular element in a collection of data, such as a list.

Two common searching techniques are:

1. Linear Search

Linear Search checks each element one by one from the beginning until the required element is found or the list ends.

It is simple and can be used with both sorted and unsorted data.

2. Binary Search

Binary Search repeatedly divides a sorted list into two halves and searches in the appropriate half.

It is faster than linear search for large sorted lists, but the data must be sorted before applying binary search.

#### Key Points:

- Searching is used to find an element.
- Linear Search checks elements one by one.
- Binary Search divides the list into halves.
- Binary Search requires sorted data.

### 35A Aim: Write a Python program to perform linear search.

```python
# Prac 35: Linear Search


# Creating a list of numbers
numbers = [10, 20, 30, 40, 50]


# Taking the number to search from the user
search = int(input("Enter the number to search: "))


# Creating a variable to check whether the number is found
found = False


# Checking each number in the list one by one
for number in numbers:

    # Check if the current number is equal to the search number
    if number == search:

        # The number is found
        found = True

        # Stop the loop
        break


# Check whether the number was found
if found:

    # Display the result
    print("Number found")

else:

    # Display the result if number is not found
    print("Number not found")
```

It checks the elements one by one.

# 35B Binary Search
### Aim: Write a Python program to perform binary search.

- Important: Binary search requires the list to be sorted.

```python
# Prac 35: Binary Search


# Creating a sorted list of numbers
numbers = [10, 20, 30, 40, 50]


# Taking the number to search from the user
search = int(input("Enter the number to search: "))


# Starting position of the list
low = 0


# Ending position of the list
high = len(numbers) - 1


# Variable to check whether the number is found
found = False


# Continue searching while low is less than or equal to high
while low <= high:

    # Find the middle position
    middle = (low + high) // 2


    # Check if the middle element is the search number
    if numbers[middle] == search:

        # Number is found
        found = True

        # Stop the loop
        break


    # If search number is greater than middle element
    elif search > numbers[middle]:

        # Search in the right half
        low = middle + 1


    # If search number is smaller than middle element
    else:

        # Search in the left half
        high = middle - 1


# Check whether the number was found
if found:

    # Display the result
    print("Number found")

else:

    # Display the result
    print("Number not found")
```

# Sorting Algorithms
## Concept

Sorting is the process of arranging data in a particular order, usually ascending or descending order.

Common sorting algorithms include:

1. Bubble Sort

Bubble Sort compares adjacent elements and swaps them if they are in the wrong order. This process is repeated until the list is sorted.

2. Selection Sort

Selection Sort repeatedly finds the smallest element from the unsorted part of the list and places it in the correct position.

3. Insertion Sort

Insertion Sort builds the sorted list one element at a time. Each new element is inserted into its correct position in the already sorted portion.

Key Points:

- Sorting arranges data in a specific order.
- Bubble Sort → Compares adjacent elements.
- Selection Sort → Selects the smallest element.
- Insertion Sort → Inserts an element into its correct position.
- These algorithms are easy to understand and useful for learning basic sorting concepts.

## Bubble Sort
###Aim: Write a Python program to perform Bubble Sort.

```python
# Prac 36: Bubble Sort


# Creating a list of numbers
numbers = [50, 20, 40, 10, 30]


# Finding the number of elements in the list
n = len(numbers)


# Outer loop
for i in range(n):

    # Inner loop
    for j in range(0, n - i - 1):

        # Compare two nearby elements
        if numbers[j] > numbers[j + 1]:

            # Swap the elements
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]


# Display the sorted list
print("Sorted list:", numbers)
```

## Selection Sort
### Aim: Write a Python program to perform Selection Sort.

```python
# Prac 36: Selection Sort


# Creating a list of numbers
numbers = [50, 20, 40, 10, 30]


# Finding the number of elements
n = len(numbers)


# Outer loop
for i in range(n):

    # Assume the current position has the smallest value
    minimum = i


    # Search for the smallest element
    for j in range(i + 1, n):

        # Compare the current element with the minimum element
        if numbers[j] < numbers[minimum]:

            # Update the position of the smallest element
            minimum = j


    # Swap the smallest element with the current element
    numbers[i], numbers[minimum] = numbers[minimum], numbers[i]


# Display the sorted list
print("Sorted list:", numbers)
```

## Insertion Sort
### Aim: Write a Python program to perform Insertion Sort.

```python
# Prac 36: Insertion Sort


# Creating a list of numbers
numbers = [50, 20, 40, 10, 30]


# Start from the second element
for i in range(1, len(numbers)):

    # Store the current element
    key = numbers[i]


    # Start comparing with the previous element
    j = i - 1


    # Move elements that are greater than key
    while j >= 0 and numbers[j] > key:

        # Move the larger element one position to the right
        numbers[j + 1] = numbers[j]

        # Move to the previous element
        j = j - 1


    # Insert the key at the correct position
    numbers[j + 1] = key


# Display the sorted list
print("Sorted list:", numbers)
```
